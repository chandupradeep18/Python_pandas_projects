```python
print("Genre distribution:")
print(bestseller['Genre'].value_counts())
print("\nTop 10 authors by appearances:")
print(bestseller['Author'].value_counts().head(10))
```

```python
print(f"Min User Rating : {bestseller['User Rating'].min()}")
print(f"Max User Rating : {bestseller['User Rating'].max()}")
print(f"Min Price : {bestseller['Price'].min()}")
print(f"Max Max : {bestseller['Price'].max()}")
print(f"Total Reviews : {bestseller['Reviews'].sum()}")
print(f"Least Reviews : {bestseller['Reviews'].min()}")
print(f"Hightest Reviews : {bestseller['Reviews'].max()}")
bestseller.nunique()
```

```python
# Cell 7: Investigate duplicates
print("Jeff Kinney's books:")
print(bestseller[bestseller['Author'] == 'Suzanne Collins'][['Name', 'Year', 'Price']])
print(bestseller[bestseller['Author'] == 'Jeff Kinney'][['Name', 'Year', 'Price']])
print(f"\nDuplicate rows (same Name + Author): {bestseller.duplicated(subset=['Name', 'Author']).sum()}")
```

```python
bestseller_clean=bestseller.drop_duplicates(subset=['Name','Author'])
print(bestseller.shape)
print(bestseller_clean.shape)
```

```python
print(bestseller_clean[bestseller_clean['Author'] == 'Suzanne Collins'][['Name', 'Year', 'Price']])
```

```python
bestseller_clean=bestseller_clean.rename(columns={
    'User Rating':'rating',
    'Reviews':'reviews',
    'Price': 'price',
    'Year': 'year',
    'Genre': 'genre',
    'Name': 'title',
    'Author': 'author'
})
bestseller_clean.columns.tolist()
```

```python
print("Top 10 authors (unique books):")
print(bestseller_clean['author'].value_counts().head(10))
```

```python
high_rated_fiction = bestseller_clean[(bestseller_clean['genre']=='Fiction') & (bestseller_clean['rating']>4.7)]
print(len(high_rated_fiction))
high_rated_fiction
```

```python
cheap_great_fiction=bestseller_clean[(bestseller_clean['genre']=='Fiction') & (bestseller_clean['rating']>4.7) & (bestseller_clean['price']<10)]
print(len(cheap_great_fiction))
cheap_great_fiction
```

```python
bestseller_clean.nlargest(10,'reviews')[['title', 'author', 'reviews', 'rating']]
```

```python
print(bestseller_clean.groupby('author')[['rating', 'price', 'reviews']].mean())
```

```python
print("\nAuthors by unique bestselling books:")
print(bestseller_clean.groupby('author')['title'].count().sort_values(ascending=False).head(10))
```

```python
bestseller_clean['price'].plot(kind='hist', bins=15, title='Price Distribution of Bestselling Books',xlabel='Price ($)', ylabel='Count')
plt.tight_layout()
plt.show()
```

```python
genre_stats = bestseller_clean.groupby('genre')[['rating', 'price']].mean()
genre_stats.plot(kind='bar', title='Fiction vs Non Fiction: Average Rating and Price')
plt.tight_layout()
plt.show()
```

```python
top_authors = bestseller_clean['author'].value_counts().head(10)
top_authors.plot(kind='hist', title='Top 10 Authors by Unique Bestselling Books',
                 xlabel='Number of Books')
plt.tight_layout()
plt.show()
```

```python
high_rated_fiction = bestseller_clean[(bestseller_clean['genre']=='Fiction') & (bestseller_clean['rating']>4.7)].head(10)
high_rated_fiction.plot(kind='pie', title='Top 10 Authors by Unique Bestselling Books',xlabel='Number of Books')
plt.tight_layout()
plt.show()
```
