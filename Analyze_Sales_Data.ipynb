```python
import pandas as pd
import sqlite3
conn=sqlite3.connect("tips")
```

```python
tips=pd.read_csv("https://raw.githubusercontent.com/pandas-dev/pandas/main/pandas/tests/io/data/csv/tips.csv")
print(f"{len(tips)}")
print(tips.shape)
tips.head(5)
tips.info()
```

```python
tips.to_sql("tips",conn,index=False,if_exists="replace")
```

```python
pd.read_sql_query("Select * From tips limit 5",conn)
```

```python
pd.read_sql_query("Select total_bill,tip,day from tips limit 10",conn)
```

```python
pd.read_sql_query("Select * From tips where time='Dinner'",conn)
```

```python
pd.read_sql_query("Select day,AVG(tip) as avg_tip,sum(tip),max(tip),min(tip),sum(total_bill),max(total_bill),min(total_bill),count(*) as count from tips group by day",conn)
```

```python
tips[['total_bill','tip','day']].head(4)
```

```python
tips[(tips['time']=='Dinner') & (tips['tip']>5)].head(4)
```

```python
tips.assign(tip_rate=tips['tip'] / tips['total_bill']).head()
```

```python
tips.head()
```

```python
results=tips.groupby("day")["tip"].mean().reset_index()
results.columns=['day','avg_tip']
print(results)
```

```python
customers = pd.DataFrame({
    "customer_id": [1, 2, 3, 4],
    "name": ["Alice", "Bob", "Charlie", "Diana"],
    "city": ["New York", "London", "Paris", "Tokyo"]
})

orders = pd.DataFrame({
    "order_id": [101, 102, 103, 104, 105],
    "customer_id": [1, 2, 2, 5, 3],
    "amount": [50.00, 30.00, 75.00, 20.00, 45.00]
})
print("Customers:")
print(customers)
print("\nOrders:")
print(orders)
```

```python
inner_result=pd.merge(orders,customers,on="customer_id",how="outer")
print("Inner Join Result:")
print(inner_result)
```

```python
customers.to_sql("customers",conn,index=False,if_exists="replace")
orders.to_sql("orders",conn,index=False,if_exists="replace")
sql_inner=pd.read_sql_query("select orders.order_id,orders.customer_id,orders.amount,customers.name, customers.city from orders full join customers on orders.customer_id=customers.customer_id",conn)
print("SQL INNER JOIN result:")
sql_inner.to_csv("sql_inner.csv",index=False)
print(sql_inner)
```

```python
orders_with_null = pd.DataFrame({
    "order_id": [101, 102, 103],
    "customer_id": [1, None, None],
    "amount": [50.00, 30.00, 75.00]
})

customers_with_null = pd.DataFrame({
    "customer_id": [1, None],
    "name": ["Alice", "Unknown"],
    "city": ["New York", "Unknown"]
})
merging_null_value=pd.merge(orders_with_null,customers_with_null,on="customer_id")
print("Merging with Null Values:")
print(merging_null_value)
```
